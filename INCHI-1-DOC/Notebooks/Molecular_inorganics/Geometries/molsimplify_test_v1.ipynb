```python
import os
import molSimplify

from molSimplify.Classes import mol3D
```

```python
#Example with one file

filename = "./data/cis_platin.mol"
complex_mol = mol3D()
complex_mol.readfrommol(filename)
complex_mol.get_geometry_type()
```

```python
directory = './data'

absolute_file_paths = [os.path.abspath(os.path.join(directory, f)) for f in os.listdir(directory) if os.path.isfile(os.path.join(directory, f)) and "mol" in f and not "xyz" in f]

for cur_fn in absolute_file_paths:
    
    complex_mol = mol3D()

    try:
        print("----------------------")
        print(cur_fn)
        complex_mol.readfrommol(cur_fn)
        geo_dict = complex_mol.get_geometry_type()
        if "geometry" in geo_dict:
            print(geo_dict["geometry"])
    except Exception as ex:
        print("error", str(ex))
```
